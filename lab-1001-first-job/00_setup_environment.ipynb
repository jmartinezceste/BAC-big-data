This notebook does three things, and you only need to run it once (**Run all**):

1. Creates the `lab1001` catalog with its schemas and the `landing.data` volume.
2. Copies the production reports from this repository's `data/` folder into the volume.
3. Prints the path of the SQL file you will use in the job task.

```python
%python
import os
import shutil

CATALOG = "lab1001"

# This notebook's folder is the lab folder inside your Git folder.
LAB_PATH = os.getcwd()
if not LAB_PATH.startswith("/Workspace"):
    ctx = dbutils.notebook.entry_point.getDbutils().notebook().getContext()
    LAB_PATH = "/Workspace" + os.path.dirname(ctx.notebookPath().get())

print("Lab:    ", LAB_PATH)
print("Catalog:", CATALOG)
```

## 1 · Catalog, schemas and volume

```python
%python
spark.sql(f"CREATE CATALOG IF NOT EXISTS {CATALOG}")
for schema in ["landing", "bronze"]:
    spark.sql(f"CREATE SCHEMA IF NOT EXISTS {CATALOG}.{schema}")
spark.sql(f"CREATE VOLUME IF NOT EXISTS {CATALOG}.landing.data")

display(spark.sql(f"SHOW SCHEMAS IN {CATALOG}"))
```

## 2 · Production reports into the volume

You should see **two files**: the report for Monday 19 and Tuesday 20 October.

```python
%python
source = f"{LAB_PATH}/data"
target = f"/Volumes/{CATALOG}/landing/data"

for file_name in sorted(os.listdir(source)):
    shutil.copyfile(f"{source}/{file_name}", f"{target}/{file_name}")
    print("copied", file_name)

display(dbutils.fs.ls(target))
```
